```python
# ============================================================
# CELL 2 (retry): Load Pretrained PaddleOCR, GPU
# ============================================================
from paddleocr import PaddleOCR
import time

ocr = PaddleOCR(
    use_doc_orientation_classify=False,
    use_doc_unwarping=False,
    use_textline_orientation=False,
    device='cpu',
    text_detection_model_name="PP-OCRv6_small_det",
    text_recognition_model_name="PP-OCRv6_small_rec",
)
print("PaddleOCR loaded")

import paddle
print("CUDA available:", paddle.device.is_compiled_with_cuda())
```

```python
# ============================================================
# CELL 3 (FIXED): Baseline Test — Full Receipt Photos (no fine-tuning)
# ============================================================
from pathlib import Path

IMAGE_DIR = Path("/kaggle/input/datasets/maazahmad69/inference/inference")
image_paths = sorted(list(IMAGE_DIR.glob("*.jpg")))
print(f"Found {len(image_paths)} images")

for path in image_paths:
    t0 = time.time()
    result = ocr.predict(str(path))
    t1 = time.time()

    print(f"\n=== {path.name} ({t1-t0:.2f}s) ===")
    res = result[0]
    texts = res["rec_texts"]
    scores = res["rec_scores"]
    for text, score in zip(texts, scores):
        print(f"  [{score:.2f}] {text}")
```
